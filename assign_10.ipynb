```python
import numpy as np

# Creating two arrays
arr1 = np.array([1, 4, 9, 16, 25])
arr2 = np.array([5, 2, 3, 4, 5])

# Element-wise operations
add = arr1 + arr2
sub = arr1 - arr2
mul = arr1 * arr2
div = arr1 / arr2

print("Array 1:", arr1)
print("Array 2:", arr2)
print("\nAddition:", add)
print("Subtraction:", sub)
print("Multiplication:", mul)
print("Division:", div)

# Using universal functions
sqrt = np.sqrt(arr1)
log = np.log(arr1)
exp = np.exp(arr2)

print("\nSquare root of arr1:", sqrt)
print("Natural logarithm of arr1:", log)
print("Exponential of arr2:", exp)

```
