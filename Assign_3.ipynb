```python
def process_file(input_file, output_file):
    try:
        # Open and read the input file
        with open(input_file, 'r', encoding='utf-8') as file:
            data = file.read()

        # Process data
        line_count = data.count('\n') + 1 if data else 0
        word_count = len(data.split())
        char_count = len(data)

        # Prepare the output content
        result = (
            f"Processed Data Summary:\n"
            f"-----------------------\n"
            f"Total Lines: {line_count}\n"
            f"Total Words: {word_count}\n"
            f"Total Characters: {char_count}\n"
        )

        # Write the processed data to output file
        with open(output_file, 'w', encoding='utf-8') as file:
            file.write(result)

        print("Processing complete. Results written to", output_file)

    except FileNotFoundError:
        print(f"Error: The file '{input_file}' was not found.")
    except PermissionError:
        print(f"Error: Permission denied while accessing the file.")
    except Exception as e:
        print(f"An unexpected error occurred: {e}")


# Example usage
input_file = r"C:\Users\chava\Documents\DESKTOP BACKUPS\AY 2025-26\input.txt"   # Replace with your input file path
output_file = r"C:\Users\chava\Documents\DESKTOP BACKUPS\AY 2025-26\output.txt" # Replace with desired output file path
process_file(input_file, output_file)
```
