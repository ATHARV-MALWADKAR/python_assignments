```python
import array
import time


# Integer array (type 'i' for signed integer)
arr = array.array('i', [1, 2, 3, 4, 5])

# Equivalent Python list
lst = [1, 2, 3, 4, 5]


# Insert 10 at index 2 in array
arr.insert(2, 10)
print("Array after insertion:", arr.tolist())

# Insert 10 at index 2 in list
lst.insert(2, 10)
print("List after insertion:", lst)


# Remove element 3 from array
arr.remove(3)
print("Array after deletion:", arr.tolist())

# Remove element 3 from list
lst.remove(3)
print("List after deletion:", lst)


# Find index of element 4 in array
index_arr = arr.index(4)
print("Index of 4 in array:", index_arr)

# Find index of element 4 in list
index_lst = lst.index(4)
print("Index of 4 in list:", index_lst)


print("Iterating array:")
for i in arr:
    print(i, end=' ')

print("\nIterating list:")
for i in lst:
    print(i, end=' ')


import timeit

# Setup
setup_code = '''
import array
arr = array.array('i', range(1000000))
lst = list(range(1000000))
'''

# Insertion at the beginning
arr_insert_time = timeit.timeit('arr.insert(0, -1)', setup=setup_code, number=1000)
lst_insert_time = timeit.timeit('lst.insert(0, -1)', setup=setup_code, number=1000)

print("\n\nInsertion at start (1000 times):")
print("Array:", arr_insert_time, "seconds")
print("List :", lst_insert_time, "seconds")
```
