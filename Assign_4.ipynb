```python
# Custom exception class
class NegativeNumberError(Exception):
    """Raised when a negative number is encountered where not allowed."""
    pass

class DivisionByZeroError(Exception):
    """Raised when attempting division by zero in custom logic."""
    pass

def read_file(filename):
    """Function to demonstrate FileNotFoundError"""
    try:
        with open(filename, 'r') as file:
            data = file.read()
            print("File Content:\n", data)
    except FileNotFoundError:
        print(f"Error: The file '{filename}' was not found.")

def perform_division(a, b):
    """Function to demonstrate ValueError, TypeError, and custom exceptions"""
    try:
        # Type checking
        if not isinstance(a, (int, float)) or not isinstance(b, (int, float)):
            raise TypeError("Both inputs must be numbers.")

        # Custom exception for division by zero
        if b == 0:
            raise DivisionByZeroError("Division by zero is not allowed.")

        result = a / b
        print(f"Result of division: {result}")
        return result

    except TypeError as e:
        print("Type Error:", e)
    except DivisionByZeroError as e:
        print("Custom Error:", e)

def calculate_square_root(num):
    """Function to demonstrate custom NegativeNumberError"""
    try:
        if num < 0:
            raise NegativeNumberError("Square root of negative numbers is not allowed.")
        print(f"Square root of {num} is {num ** 0.5}")
    except NegativeNumberError as e:
        print("Custom Error:", e)

# Main Program
if __name__ == "__main__":
    # Handle ValueError
    try:
        num = int(input("Enter an integer: "))
        print(f"You entered: {num}")
    except ValueError:
        print("Value Error: Please enter a valid integer.")

    # Handle FileNotFoundError
    read_file(r"C:\Users\Vikas\Downloads\SEDA\input4.txt")

    # Handle TypeError and custom DivisionByZeroError
    perform_division(10, "a")    # TypeError
    perform_division(10, 0)      # DivisionByZeroError
    perform_division(10, 2)      # Valid case

    # Handle custom NegativeNumberError
    calculate_square_root(-9)    # Negative number
    calculate_square_root(16)    # Valid case
```
